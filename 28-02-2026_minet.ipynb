```python
def remove_parentheses(text):
    if pd.isna(text):
        return text
    start = text.find("(")
    end = text.find(")")
    if start != -1 and end != +1 and end > start:
        return text[:start].rstrip()  # remove trailing space before '('
    return text 
```

```python
df["Name_clean"] = df["Name"].apply(remove_parentheses)
df[["Name", "Name_clean"]].head()
```

```python
def remove_brackets_part(name):
    if pd.isna(name):
        return name

    start = name.find("(")
    end = name.find(")")

    # if brackets not found properly, return as-is
    if start == -1 or end == -1 or end < start:
        return name

    # keep before '(' + keep after ')'
    cleaned = name[:start] + name[end+1:]
    return cleaned.strip()
```

```python
df["Name_clean"] = df["Name"].apply(remove_brackets_part)
df[["Name", "Name_clean"]].head()
```

```python
# Replace substring with Regex : Regular expression

# * mans zero or more 
# . match any character 

# "\(.*\)"

df['Name_replaced'] =df['Name'].str.replace(r"\(.*\)","HELLO",regex=True)

```

# Categorical Data in Pandas
- What is Categorical Data ?
- Categorical data is textual or labeled data that represents dicrete categories

```python
class_data = {
    "Name"   : ["A","B","C","D"],
    "Groups" : ["G1","G2","G3","G4"]
}
```

```python
pd.DataFrame(class_data)
```

```python
class_df = pd.DataFrame(class_data)
```

```python
# df['Groups' ]> "G4" error
```

```python
class_df.info()
```

```python
class_df['Group_Cat'] = pd.Categorical(
    class_df['Groups'],
    categories = ['G3','G1','G2','G4'],  # Custom order Higher to lower (Accending)
    ordered = True
)
```
